### W6A4 Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó

```python
chuoi_nhap = input("Nhập các cặp key:value: ")
cac_cap = chuoi_nhap.split()
ket_qua = {}
for cap in cac_cap:   
    key, value = cap.split(":", 1)
    if key in ket_qua:
        ket_qua[key].append(value)
    else:
        ket_qua[key] = [value]
print(ket_qua)
    
```

### W6A5 Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros': ...} (đếm số >0, <0, =0)

```python
chuoi = input("Nhập chuỗi: ")
so = list(map(int, chuoi.split()))
positive = 0
negative = 0
zeros = 0
for i in so:
    if i > 0:
        positive += 1
    elif i < 0:
        negative += 1
    else:
        zeros += 1
dap_an = {"positive": positive, "negative": negative, "zeros": zeros}
print(dap_an)
```

### W6A9 Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của key.  

```python
from collections import Counter
def tao_dict_tu_chuoi(chuoi_nhap):
    d = {}
    if not chuoi_nhap.strip(): 
        return d
    cac_cap = chuoi_nhap.split()
    for cap in cac_cap:
        key, value = cap.split(':')
        d[key] = int(value) 
    return d
print("Nhập dữ liệu theo định dạng key:value (VD: a:1 b:2)")
chuoi1 = input("Nhập dictionary 1: ")
chuoi2 = input("Nhập dictionary 2: ")
dict1 = tao_dict_tu_chuoi(chuoi1)
dict2 = tao_dict_tu_chuoi(chuoi2)
dict_gop = Counter(dict1) + Counter(dict2)
dict_ket_qua = dict(sorted(dict_gop.items()))
print("Dictionary sau khi gộp và sắp xếp:", dict_ket_qua)
```

### W6A10 Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple.

```python
def tim_cap_so(danh_sach, k):
    seen = set()       
    ket_qua = set()    
    for num in danh_sach:
        target = k - num
        if target in seen:
            ket_qua.add((target, num))
        seen.add(num)
    ket_qua_list = sorted(list(ket_qua), key=lambda x: x[0])
    return ket_qua_list
if __name__ == "__main__":
    try:
        chuoi_nhap = input("Nhập danh sách các số nguyên, cách nhau bởi khoảng trắng: ")
        danh_sach = [int(x) for x in chuoi_nhap.split()] 
        k = int(input("Nhập số nguyên dương k: "))
        if k <= 0:
            print("Lỗi: k phải là một số nguyên dương.")
        else:
            kq = tim_cap_so(danh_sach, k)
            print("Danh sách các cặp số thỏa mãn:", kq)
    except ValueError:
        print("Lỗi: Vui lòng nhập đúng định dạng số nguyên.")
```

### W6A11 [Tách số chẵn và lẻ từ tuple] Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt. 

```python
chuoi_nhap = input("Nhập các số nguyên, cách nhau bởi khoảng trắng: ")
tuple_goc = tuple(int(x) for x in chuoi_nhap.split())
danh_sach_chan = []
danh_sach_le = []
for so in tuple_goc:
    if so % 2 == 0:
        danh_sach_chan.append(so)
    else:
        danh_sach_le.append(so)
tuple_chan = tuple(danh_sach_chan)
tuple_le = tuple(danh_sach_le)
print(f"Tuple chứa các số chẵn: {tuple_chan}")
print(f"Tuple chứa các số lẻ: {tuple_le}")
```

### W6A12 [Tìm phần tử xuất hiện nhiều nhất] Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất trong số đó.

```python
chuoi_nhap = input("Nhập list các số nguyên, cách nhau bởi khoảng trắng: ")
danh_sach = [int(x) for x in chuoi_nhap.split()]
dem_tan_suat = {}
for so in danh_sach:
    if so in dem_tan_suat:
        dem_tan_suat[so] += 1
    else:
        dem_tan_suat[so] = 1
so_tot_nhat = None
so_lan_max = 0
for so, so_lan in dem_tan_suat.items():
    if so_lan > so_lan_max:
        so_lan_max = so_lan
        so_tot_nhat = so
    elif so_lan == so_lan_max:
        if so_tot_nhat is None or so < so_tot_nhat:
            so_tot_nhat = so
print(f"Số xuất hiện nhiều nhất thỏa mãn điều kiện là: {so_tot_nhat}")
```

### W6A13 [Đảo ngược key và value của dictionary] Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo một dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In ra dictionary mới.

```python
chuoi_nhap = input()
danh_sach_tu = chuoi_nhap.split()
dict_dao_nguoc = {}
for i in range(0, len(danh_sach_tu), 2):
    if i + 1 < len(danh_sach_tu):
        key_cu = danh_sach_tu[i]
        value_cu = danh_sach_tu[i + 1]
        dict_dao_nguoc[value_cu] = key_cu
print(dict_dao_nguoc)
```

### W6A15 [Lọc dictionary theo giá trị] Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp key value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới.

```python
chuoi_nhap = input()
danh_sach_tu = chuoi_nhap.split()
k = int(input())
dict_goc = {}
for i in range(0, len(danh_sach_tu), 2):
    if i + 1 < len(danh_sach_tu):
        key = danh_sach_tu[i]
        value = int(danh_sach_tu[i + 1]) 
        dict_goc[key] = value
dict_loc = {key: val for key, val in dict_goc.items() if val > k}
print(dict_loc)
```
