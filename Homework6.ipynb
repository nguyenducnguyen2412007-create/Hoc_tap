### W6A16 Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải 

```python
n, m = map(int, input("Nhập số hàng n và số cột m (cách nhau bởi dấu cách): ").split())
ma_tran = []
print(f"Nhập {n} hàng, mỗi hàng gồm {m} số nguyên:")
for _ in range(n):
    hang = list(map(int, input().split()))
    ma_tran.append(hang)
print("\nMa trận vừa nhập:")
for hang in ma_tran:
    for so in hang:
        print(f"{so:>4}", end="")
    print()  
```

### W6A17 Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng 

```python
def in_duong_cheo():
    n = int(input("Nhập kích thước n của mảng vuông: "))
    matrix = []
    print(f"Nhập các phần tử của mảng {n}x{n} (mỗi hàng cách nhau bởi khoảng trắng):")
    for i in range(n):
        row = list(map(int, input(f"Nhập hàng {i}: ").split()))
        matrix.append(row)
    main_diagonal = []
    secondary_diagonal = []
    for i in range(n):
        main_diagonal.append(matrix[i][i])
        secondary_diagonal.append(matrix[i][n - 1 - i])
    print("Đường chéo chính:", main_diagonal)
    print("Đường chéo phụ:", secondary_diagonal)
in_duong_cheo()
```

### W6A18 Nhập vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột k (hàng k)

```python
def tinh_tong_hang_cot_k():
    try:
        n, m, k = map(int, input("Nhập n, m, k: ").split())
    except ValueError:
        print("Vui lòng nhập đúng định dạng: n m k (cách nhau bởi khoảng trắng).")
        return
    matrix = []
    print(f"Nhập dữ liệu cho mảng {n} hàng, {m} cột:")
    for i in range(n):
        row = list(map(int, input(f"Nhập hàng {i}: ").split()))
        if len(row) != m:
            print(f"Lỗi: Vui lòng nhập đúng {m} phần tử cho hàng {i}.")
            return
        matrix.append(row)
    print("\n--- Kết Quả ---")
    if 0 <= k < n:
        tong_hang_k = sum(matrix[k])
        print(f"Tổng các phần tử trên hàng {k} là: {tong_hang_k}")
    else:
        print(f"Hàng {k} không tồn tại trong mảng (mảng chỉ có các hàng từ 0 đến {n-1}).")
    if 0 <= k < m:
        tong_cot_k = sum(matrix[i][k] for i in range(n))
        print(f"Tổng các phần tử trên cột {k} là: {tong_cot_k}")
    else:
        print(f"Cột {k} không tồn tại trong mảng (mảng chỉ có các cột từ 0 đến {m-1}).")
tinh_tong_hang_cot_k()
```

### W6A19 Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất trong list 

```python
def vị_tri_lon_nhat(lst):
    if not lst:
        return -1  
    max_value = lst[0]
    max_index = 0
    for i in range(1, len(lst)):
        if lst[i] > max_value:
            max_value = lst[i]
            max_index = i
    return max_index
a = input("Nhập một danh sách số nguyên, cách nhau bằng dấu cách: ")
lst = list(map(int, a.split()))
index = vị_tri_lon_nhat(lst)
if index != -1:
    print(f"Vị trí của phần tử lớn nhất là: {index}")
else:   
    print("Danh sách rỗng.")
```

### W6A20 Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1 

```python
def vi_tri_dau_tien(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1
a = input("Nhập vào một chuỗi: ")
k = input("Nhập vào một số nguyên: ")
lst = a.split()
result = vi_tri_dau_tien(lst, k)    
if result != -1:
    print(f"Vị trí đầu tiên của {k} trong chuỗi là: {result}")
else:
    print(f"{k} không có trong chuỗi.")
```

### W6A21 Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
chuoi_dau_vao = input("Nhập các chuỗi (cách nhau bởi dấu cách): ")
danh_sach_chuoi = chuoi_dau_vao.split()
tu_dien_vi_tri = {}
for i, tu in enumerate(danh_sach_chuoi):
    if tu not in tu_dien_vi_tri:
        tu_dien_vi_tri[tu] = [i]
    else:
        tu_dien_vi_tri[tu].append(i)
for tu in tu_dien_vi_tri:
    tu_dien_vi_tri[tu] = tuple(tu_dien_vi_tri[tu])
print("\nKết quả ánh xạ Dictionary:")
print(tu_dien_vi_tri)
```

### W6A23 Viết hàm xóa đi các số <x có trong mảng 

```python
def xoa_nho_hon_x(lst, x):
    return [so for so in lst if so >= x]
chuoi_nhap = input("Nhập một mảng các số nguyên (cách nhau bởi dấu cách): ")
danh_sach_so = [int(so) for so in chuoi_nhap.split()]
x = int(input("Nhập số x (các số nhỏ hơn x sẽ bị xóa): "))
ket_qua = xoa_nho_hon_x(danh_sach_so, x)
print("-" * 40)
print(f"Mảng ban đầu: {danh_sach_so}")
print(f"Mảng sau khi xóa các số < {x}: {ket_qua}")
```

### W6A24 Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9 

```python
def day_0_ve_dau(lst):
    so_0 = [x for x in lst if x == 0]
    so_khac_0 = [x for x in lst if x != 0]
    return so_0 + so_khac_0
chuoi_nhap = input("Nhập danh sách các số (cách nhau bởi dấu cách): ")
danh_sach = [int(so) for so in chuoi_nhap.split()]
print("-" * 40)
print(f"Mảng ban đầu:  {danh_sach}")
print(f"Đẩy 0 về đầu:  {day_0_ve_dau(danh_sach)}")
```

### W6A25 Cho dãy chỉ gồm N giá trị từ 0 - 9 . Kiểm tra xem giá trị nào không có trong mảng 

```python
def tim_so_thieu(arr):
    all_digits = set(range(10))
    missing_digits = all_digits - set(arr)
    return sorted(list(missing_digits))
mang_dau_vao = [int(x) for x in input("Nhập một mảng số (cách nhau bằng dấu phẩy): ").split(",")]
ket_qua = tim_so_thieu(mang_dau_vao)
print("Các giá trị không có trong mảng:", ket_qua)
```
