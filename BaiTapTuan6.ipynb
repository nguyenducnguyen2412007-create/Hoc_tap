### W6A4 Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó

```python
chuoi = input("Nhập chuỗi: ")
ket_qua_chuoi = {}
tach = chuoi.split()
for i in tach:
        key, value = i.split(':')
        if key not in ket_qua_chuoi:
            ket_qua_chuoi[key] = []
        ket_qua_chuoi[key].append(value)
print(ket_qua_chuoi)
```

### W6A5 Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros': ...} (đếm số >0, <0, =0)

### W6A9 Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của key.  

```python
def nhap_dictionary(ten_dict):
    d = {}
    print(f"--- Nhập {ten_dict} ---")
    n = int(input("Nhập số lượng phần tử: "))
    for i in range(n):
        key = input(f"Nhập key thứ {i+1}: ")
        value = int(input(f"Nhập giá trị cho key '{key}': "))
        d[key] = value
    return d
def gop_va_sap_xep_dict():
    dict1 = nhap_dictionary("Dictionary 1")
    dict2 = nhap_dictionary("Dictionary 2")
    merged_dict = dict1.copy() 
    for key, value in dict2.items():
        if key in merged_dict:
            merged_dict[key] += value 
        else:
            merged_dict[key] = value  
    print("\n--- Kết quả Dictionary sau khi gộp ---")
    for key in sorted(merged_dict.keys()):
        print(f"'{key}': {merged_dict[key]}")
if __name__ == "__main__":
    gop_va_sap_xep_dict()
```

### Q6A10 Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple. 

```python
def tim_cap_so_tong_k():
    chuoi_nhap = input("Nhập danh sách các số nguyên (cách nhau bởi khoảng trắng): ")
    danh_sach = [int(x) for x in chuoi_nhap.split()]
    k = int(input("Nhập số nguyên dương k: "))
    ket_qua_set = set()
    n = len(danh_sach)
    for i in range(n):
        for j in range(i + 1, n):
            if danh_sach[i] + danh_sach[j] == k:
                ket_qua_set.add((danh_sach[i], danh_sach[j]))
    ket_qua_list = sorted(list(ket_qua_set), key=lambda x: x[0])
    print("\n--- Kết quả ---")
    print(ket_qua_list)
if __name__ == "__main__":
    tim_cap_so_tong_k()
```

### W6A11[Tách số chẵn và lẻ từ tuple] 
### Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt. 

```python
def tach_chan_le():
    chuoi_nhap = input("Nhập các số nguyên (cách nhau bởi khoảng trắng): ")
    tuple_goc = tuple(int(x) for x in chuoi_nhap.split())
    tuple_chan = tuple(x for x in tuple_goc if x % 2 == 0)
    tuple_le = tuple(x for x in tuple_goc if x % 2 != 0)
    print("\n--- Kết quả ---")
    print(tuple_chan)
    print(tuple_le)
if __name__ == "__main__":
    tach_chan_le()
```

### W6A12 [Tìm phần tử xuất hiện nhiều nhất] 
### Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất trong số đó. 

```python
from collections import Counter
def so_xuat_hien_nhieu_nhat():
    chuoi_nhap = input("Nhập danh sách các số nguyên (cách nhau bởi khoảng trắng): ")
    lst = [int(x) for x in chuoi_nhap.split()]
    dem = Counter(lst)
    max_xuat_hien = max(dem.values())
    so_nhieu_nhat = [so for so, xuat_hien in dem.items() if xuat_hien == max_xuat_hien]
    return min(so_nhieu_nhat)
if __name__ == "__main__":
    print(f"Số xuất hiện nhiều nhất: {so_xuat_hien_nhieu_nhat()}")
```

### W6A13 [Đảo ngược key và value của dictionary] 
### Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo một dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In ra dictionary mới. 

```python
def dao_nguoc_dictionary():
    n = int(input("Nhập số lượng phần tử của dictionary: "))
    dict_goc = {}
    for i in range(n):
        key = input(f"Nhập key thứ {i+1}: ")
        value = input(f"Nhập value cho key '{key}': ")
        dict_goc[key] = value
    dict_dao_nguoc = {v: k for k, v in dict_goc.items()}
    print("\n--- Dictionary ban đầu ---")
    print(dict_goc)
    print("\n--- Dictionary sau khi đảo ngược ---")
    print(dict_dao_nguoc)
if __name__ == "__main__":
    dao_nguoc_dictionary()
```

### W6A14 [Giao của hai danh sách] 
### Nhập vào hai dòng, mỗi dòng là một danh sách các số nguyên. Hãy tìm các phần tử chung (giao) của hai danh sách này và in ra một list mới chứa các phần tử chung đó, không trùng lặp. 

```python
def tim_giao_hai_danh_sach():
    chuoi_1 = input("Nhập danh sách 1 (cách nhau bởi khoảng trắng): ")
    chuoi_2 = input("Nhập danh sách 2 (cách nhau bởi khoảng trắng): ")
    list1 = [int(x) for x in chuoi_1.split()]
    list2 = [int(x) for x in chuoi_2.split()]
    set_giao = set(list1) & set(list2)
    ket_qua = list(set_giao)
    print("\n--- Kết quả ---")
    print(f"Các phần tử chung: {ket_qua}")
if __name__ == "__main__":
    tim_giao_hai_danh_sach()
```

### W6A15 [Lọc dictionary theo giá trị] 
### Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp key-value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới.

```python
def loc_dictionary_theo_gia_tri():
    n = int(input("Nhập số lượng phần tử của dictionary: "))
    dict_goc = {}
    for i in range(n):
        key = input(f"Nhập key thứ {i+1} (chuỗi): ")
        value = int(input(f"Nhập value cho key '{key}' (số nguyên): ")) 
        dict_goc[key] = value
    k = int(input("Nhập số nguyên k: "))
    dict_moi = {key: value for key, value in dict_goc.items() if value > k}
    print("\n--- Dictionary ban đầu ---")
    print(dict_goc)
    print(f"\n--- Dictionary sau khi lọc (các giá trị > {k}) ---")
    print(dict_moi)
if __name__ == "__main__":
    loc_dictionary_theo_gia_tri()
```
